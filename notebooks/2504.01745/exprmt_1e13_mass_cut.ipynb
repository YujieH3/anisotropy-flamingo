```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
plt.style.use('./nice.mplstyle')

# Load the data
catalogue = pd.read_csv("/cosma/home/do012/dc-he4/anisotropy-flamingo/data/obs/eRASS1-full-final.txt", sep="\t")

# Convert flux from 1e-14 units to erg/s/cm² and apply flux cut
flux_threshold = 5e-12 / 1e-14  # 500 in catalogue units
filtered = catalogue[catalogue["Fx(0.2-2.3,1e-14)"] > flux_threshold]
print(len(catalogue))
print(len(filtered))

# Plot mass distribution (M500 in units of 1e13 Msun)
plt.figure(figsize=(5, 3))
plt.hist(np.log10(filtered["M500(1e13)"]), bins=15)
plt.xlabel(r"$\log[M_{500}/10^{13}M_\odot]$")
plt.ylabel("Number of Clusters")
# plt.title("Cluster Mass Distribution ($F_X > 5 \\times 10^{-12}$ erg/s/cm²)")
# plt.grid(True)
plt.tight_layout()

plt.savefig('./to_paper/discuss_1e13_mass_cut_eRASS1.png', bbox_inches='tight', dpi=200)
plt.show()
```
