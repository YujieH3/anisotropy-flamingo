```python
%load_ext autoreload
%autoreload 2
import matplotlib.pyplot as plt
import numpy as np
```

```python
data = np.loadtxt('sb_30logbins.txt')
```

```python
np.shape(data)
```

```python
Nbin = 30
logR = np.linspace(-2, 0, Nbin)
R = 10**logR

upp = np.percentile(data, q=84, axis=0)
mid = np.percentile(data, q=50, axis=0)
low = np.percentile(data, q=16, axis=0)
```

```python
fig, ax = plt.subplots(figsize=(6,5), dpi=300)
ax.fill_between(R[:-1], low, upp, color='grey')
ax.plot(R[:-1], mid, color='black')
ax.set(ylabel='SB', xlabel='r/r500c')
ax.set_xscale('log')
ax.set_yscale('log')
plt.show()
```
