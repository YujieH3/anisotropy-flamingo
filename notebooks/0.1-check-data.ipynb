```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
def read_sample(path):
    df = pd.read_csv(path, sep=r'\s+', engine='python')
    return df.rename(columns={'#Name': 'Name'})
```

```python
data = read_sample('../data/Sample-Lx-Tx-CS.txt')
data
```

```python
# Quick sanity check
plt.figure()
plt.subplot(projection="hammer")
plt.scatter(x=data['Glon']*np.pi/180 - np.pi, y=data['Glat']*np.pi/180)
plt.grid(True)
```

```python
# Shuffle experiment
data[['Glon', 'Glat']] = data[['Glon', 'Glat']].sample(frac=1, replace=False).values
```
