```python
import pandas as pd
import numpy as np
import os

data_dir = '/cosma8/data/do012/dc-he4/analysis'

for lc in range(1728):
    lc00 = f'{lc:04d}'
    # flag filenames
    lcpath = f'{data_dir}/lc{lc00}/'

    files = os.listdir(lcpath)
    for file in files:
        if 'samples_in_lightcone' in file:
            # print(file)
            df = pd.read_csv(os.path.join(lcpath, file))
            df_null_counts = np.sum(df.isnull())
            if (df_null_counts).any():
                print(f'Lightcone {lc00} has at least one empty entry in file {file}.')
    # break
    
```
